# AWS Route 53 SME Knowledge Base Notebook
This notebook provides python helpers to load, search, and query the AWS Route 53 SME dataset in Google Colab.

```python
# Cell 1: Load and parse Route 53 Knowledge Base into DataFrame
import pandas as pd
import re

# Upload or read Master file
file_path = "00_Route53_SME_ALL_IN_ONE_MASTER.md"
try:
    with open(file_path, "r", encoding="utf-8") as f:
        text = f.read()
except FileNotFoundError:
    print("Please upload 00_Route53_SME_ALL_IN_ONE_MASTER.md to Colab runtime files.")
    text = ""

# Split modules
sections = text.split("# FILE: ")
data = []
for sec in sections[1:]:
    lines = sec.split("
", 1)
    filename = lines[0].strip()
    content = lines[1] if len(lines) > 1 else ""
    data.append({"filename": filename, "content": content, "char_len": len(content)})

df = pd.DataFrame(data)
print(f"Loaded {len(df)} documents into DataFrame.")
df.head()
```

```python
# Cell 2: Keyword Search Function across Route 53 SME KB
def search_kb(query, max_results=5):
    results = df[df["content"].str.contains(query, case=False, na=False)]
    print(f"Found {len(results)} matching modules for query: '{query}'")
    for idx, row in results.head(max_results).iterrows():
        print(f"
--- [Matched: {row['filename']}] ---")
        # Print snippet
        match_pos = row["content"].lower().find(query.lower())
        start = max(0, match_pos - 100)
        end = min(len(row["content"]), match_pos + 300)
        print("... " + row["content"][start:end].replace("
", " ") + " ...")

# Example search:
# search_kb("Resolver Endpoint")
```
